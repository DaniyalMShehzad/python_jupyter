## Write a program that takes a sentence, an old word, and a new word as input.
Replace all occurrences of the old word with the new word.
Print the updated sentence.

```python
sentence = input("Enter a sentence: ")
old_word = input("Enter the word to replace: ")
new_word = input("Enter the new word: ")
updated_sentence = sentence.replace(old_word, new_word)
print("Updated sentence:", updated_sentence)
```

## A store offers a 10% discount on a product that costs 1500. Write a Python expression to calculate the discounted price using the appropriate arithmetic operator. Ask the user to enter the product price and discount rate.

```python
price = float(input("Enter product price: "))
discount = float(input("Enter discount rate (%): "))
discount_amount = price * (discount / 100)
final_price = price - discount_amount
print("Discounted price:", final_price)
```

## You have some apples. Ask the user to enter the total number of apples they have and the number of apples they give to a friend. Write a Python program to calculate how many apples are left.
Which operator will you use to find the remaining apples?

```python
total_apples = int(input("Enter total number of apples: "))
given_apples = int(input("Enter apples given to friend: "))
remaining_apples = total_apples - given_apples
print("Remaining apples:", remaining_apples)
```

## Ask the user to enter the ages of two friends. Write a Python program to check who is older.
Which comparison operator will you use to find out who is older?

```python
age1 = int(input("Enter the age of first friend: "))
age2 = int(input("Enter the age of second friend: "))
if age1 > age2:
    print("First friend is older.")
elif age2 > age1:
    print("Second friend is older.")
else:
    print("Both friends are the same age.")
```

## Ask the user to enter a product price (float). Print:
"The price of the product is $"
Format the number to 2 decimal places.

```python
price = float(input("Enter product price: "))

print(f"The price of the product is ${price:.2f}")
```

## Ask the user to enter a city and temperature. Print:
"The temperature in  is °C today."

```python
city = input("Enter city: ")
temperature = float(input("Enter temperature: "))

print("The temperature in {} is {}°C today.".format(city, temperature))
```

## Ask the user to enter:
Speed of the car (km/h)
Time traveled (hours)
Calculate and print the distance traveled.

```python
speed = float(input("Enter car speed (km/h): "))
time = float(input("Enter time traveled (hours): "))
distance = speed * time
print(f"Distance traveled: {distance} km")
```
